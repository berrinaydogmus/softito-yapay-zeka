```python
#8
def sayi_durumu(sayi):
    if sayi > 0:
        return "Pozitif"
    elif sayi < 0:
        return "Negatif"
    else:
        return "Sıfır"

print(sayi_durumu(-5))

def en_buyuk_bul(a, b, c):
    en_buyuk = a
    if b > en_buyuk:
        en_buyuk = b
    if c > en_buyuk:
        en_buyuk = c
    return en_buyuk

print(en_buyuk_bul(4, 9, 6))
```

```python
#11
import builtins
max = builtins.max
min = builtins.min
not1 = float(input("1. ders notunu girin: "))
not2 = float(input("2. ders notunu girin: "))
not3 = float(input("3. ders notunu girin: "))
not4 = float(input("4. ders notunu girin: "))

ortalama = (not1 + not2 + not3 + not4) / 4

if ortalama >= 90:
    harf_notu = "AA"
elif ortalama >= 80:
    harf_notu = "BA"
elif ortalama >= 70:
    harf_notu = "BB"
elif ortalama >= 60:
    harf_notu = "CB"
elif ortalama >= 50:
    harf_notu = "CC"
else:
    harf_notu = "FF"

print("Ortalama:", ortalama, "Harf notu:", harf_notu)

notlar = [not1, not2, not3, not4]
print("En yüksek not:", max(notlar))
print("En düşük not:", min(notlar))
```

```python
#12
sepet = []
urun_sayisi = int(input("Kaç ürün eklemek istersiniz? "))

for i in range(urun_sayisi):
    urun_adi = input("Ürün adını girin: ")
    fiyat = float(input("Ürün fiyatını girin: "))
    sepet.append((urun_adi, fiyat))

print("100 TL üzeri ürünler:")
for urun_adi, fiyat in sepet:
    if fiyat > 100:
        print(urun_adi, fiyat)

toplam = 0
for urun_adi, fiyat in sepet:
    toplam += fiyat

print("Toplam tutar:", toplam)

if toplam > 500:
    print("%10 indirim kazandınız")
```

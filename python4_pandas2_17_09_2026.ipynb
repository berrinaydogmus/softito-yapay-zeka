```python
#Bolum9 Merge
musteriler = pd.DataFrame({
    "musteri_id": [1, 2, 3, 4],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya"]
})

siparisler = pd.DataFrame({
    "musteri_id": [1, 2, 5],
    "urun": ["Kulaklık", "Şarj Kablosu", "Powerbank"],
    "tutar": [450, 90, 320]
})
```

```python
inner_sonuc = pd.merge(musteriler, siparisler, on="musteri_id", how="inner")
print(inner_sonuc)
```

```python
left_sonuc = pd.merge(musteriler, siparisler, on="musteri_id", how="left")
print(left_sonuc)
```

```python
outer_sonuc = pd.merge(musteriler, siparisler, on="musteri_id", how="outer")
print(outer_sonuc)
```

```python
#Bolum10 Concat
hafta1 = pd.DataFrame({"gun": ["Pzt", "Sal"], "adet": [20, 35]})
hafta2 = pd.DataFrame({"gun": ["Çar", "Per"], "adet": [28, 40]})

birlesik = pd.concat([hafta1, hafta2], ignore_index=True)
print(birlesik)
```

```python
#Bolum11 GroupBy
df_calisan = pd.DataFrame({
    "sube": ["Kadıköy", "Beşiktaş", "Kadıköy", "Şişli", "Beşiktaş"],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya", "Gökhan"],
    "satis": [12000, 18500, 9500, 21000, 16000]
})
```

```python
print(df_calisan.groupby("sube")["satis"].sum())
```

```python
print(df_calisan.groupby("sube")["satis"].mean())
```

```python
print(df_calisan.groupby("sube")["satis"].agg(["sum", "mean", "count"]))
```

```python
#Bolum12 Pivot Table
df_satis = pd.DataFrame({
    "sehir": ["Bursa", "Adana", "Antalya", "Bursa", "Adana"],
    "kategori": ["Elektronik", "Giyim", "Elektronik", "Giyim", "Elektronik"],
    "satis": [5000, 3200, 4100, 2800, 3900]
})
```

```python
pivot1 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis")
print(pivot1)
```

```python
pivot2 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis", aggfunc="sum", fill_value=0)
print(pivot2)
```

```python
#Bolum15 Matplotlib ile Grafik Çizme
df_gelir = pd.DataFrame({
    "ceyrek": ["Ç1", "Ç2", "Ç3", "Ç4"],
    "gelir": [42000, 55000, 48000, 61000],
    "gider": [30000, 38000, 33000, 40000]
})
```

```python
df_gelir.plot(x="ceyrek", y="gelir", kind="line", marker="o")
plt.title("Çeyreklik Gelir")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.show()
```

```python
df_gelir.plot(x="ceyrek", y="gelir", kind="bar", color="steelblue")
plt.title("Çeyreklik Gelir")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.show()
```

```python
df_gelir.plot(y="gelir", kind="pie", labels=df_gelir["ceyrek"], autopct="%1.1f%%")
plt.title("Çeyreklere Göre Gelir Payı")
plt.ylabel("")
plt.show()
```

```python
df_gelir.plot(x="gelir", y="gider", kind="scatter")
plt.title("Gelir - Gider İlişkisi")
plt.xlabel("Gelir (TL)")
plt.ylabel("Gider (TL)")
plt.show()
```

```python
#Bonus Kapsamlı Uygulama
kurslar = pd.DataFrame({
    "kurs_adi": ["Python 101", "Photoshop Temelleri", "SEO Stratejileri", "React ile Web",
                 "Illustrator Giriş", "Sosyal Medya Pazarlama", "Java Temelleri", "UI/UX Tasarım"],
    "kategori": ["Yazılım", "Tasarım", "Pazarlama", "Yazılım",
                 "Tasarım", "Pazarlama", "Yazılım", "Tasarım"],
    "satis_adedi": [120, 45, 60, 95, 30, 75, 55, 40]
})
print(kurslar)
```

```python
kategori_toplam = kurslar.groupby("kategori")["satis_adedi"].sum()
print(kategori_toplam)
```

```python
en_cok_satan = kurslar.sort_values("satis_adedi", ascending=False).head(3)
print(en_cok_satan)
```

```python
kategori_toplam.plot(kind="bar", color="steelblue")
plt.title("Kategoriye Göre Toplam Satış")
plt.xlabel("Kategori")
plt.ylabel("Toplam Satış Adedi")
plt.show()
```

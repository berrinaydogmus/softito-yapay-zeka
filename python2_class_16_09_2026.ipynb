```python
#1
class Araba:
  def __init__(self,marka, model, hiz):
    self.marka = marka
    self.model = model
    self.hiz = hiz

  def hizlandir(self, artis):
    self.hiz += artis

  def bilgi_goster(self):
    print(self.marka, self.model, self.hiz, "km/s")

araba1 = Araba("Toyota", "Corolla", 0)
araba1.bilgi_goster()
araba1.hizlandir(40)
araba1.bilgi_goster()
```

```python
#2
class Dikdortgen:
  def __init__(self,uzunluk, genislik):
    self.uzunluk = uzunluk
    self.genislik = genislik

  def alan_hesapla(self):
    return self.uzunluk * self.genislik

  def cevre_hesapla(self):
    return 2 * (self.uzunluk + self.genislik)

dikdortgen1 = Dikdortgen(5, 3)
print("Alan: ", dikdortgen1.alan_hesapla())
print("Çevre: ", dikdortgen1.cevre_hesapla())
```

```python
#3
class Kitap:
  def __init__(self, ad, yazar, sayfa_sayisi):
    self.ad = ad
    self.yazar = yazar
    self.sayfa_sayisi = sayfa_sayisi

  def ozet(self):
    return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"

kitap1 = Kitap("Suç ve Ceza", "Dostoyevski", 671)
print(kitap1.ozet())
```

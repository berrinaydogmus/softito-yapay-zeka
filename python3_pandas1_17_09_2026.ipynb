```python
#7 Filtreleme
kisiler = pd.DataFrame({
    "isim": ["Ayşe", "Mehmet", "Zeynep", "Can", "Elif", "Burak", "Deniz", "Selin"],
    "yas": [22, 27, 19, 35, 41, 24, 30, 17],
    "maas": [15000, 25000, 12000, 32000, 28000, 21000, 26000, 10000],
    "sehir": ["İstanbul", "Ankara", "İstanbul", "İzmir", "Bursa", "İstanbul", "Ankara", "İzmir"]
})
print(kisiler)
```

```python
print(kisiler[kisiler["yas"] > 25])
```

```python
print(kisiler[(kisiler["maas"] >= 20000) & (kisiler["maas"] <= 30000)])
```

```python
print(kisiler[kisiler["sehir"] == "İstanbul"])
```

```python
print(kisiler[(kisiler["yas"] < 20) | (kisiler["yas"] > 40)])
```

```python
#Bonus
sinif = pd.DataFrame({
    "isim": ["Ayşe", "Mehmet", "Zeynep", "Can", "Elif", "Burak"],
    "yas": [21, 23, 20, 22, 24, 19],
    "not1": [80, 60, 90, 55, 75, 40],
    "not2": [70, 65, 85, 50, 80, 45],
    "not3": [90, 70, 95, 60, 85, 50],
    "sehir": ["İstanbul", "Ankara", "İzmir", "Bursa", "İstanbul", "Ankara"]
})
print(sinif)
```

```python
sinif["ortalama"] = (sinif["not1"] + sinif["not2"] + sinif["not3"]) / 3
print(sinif)
```

```python
basarili = sinif[sinif["ortalama"] > 70]
print(basarili)
```

```python
print(basarili.sort_values("yas"))
```
